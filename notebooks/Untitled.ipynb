```python
import json
import numpy as np
import matplotlib.pyplot as plt

with open("/home/norman/ml-tau/ml-tau-en-reg/training-outputs/260209_jet_reg/jet_regression/ParticleTransformer/history.json") as f:
    hist = json.load(f)

epochs = np.arange(len(hist["IQR_validation"]))

iqr_val = np.array(hist["IQR_validation"])
median_val = np.array(hist["median_validation"])

resolution_val = iqr_val / median_val

plt.figure(figsize=(7, 5))
plt.plot(epochs, resolution_val, label="Validation")
plt.xlabel("Epoch")
plt.ylabel("IQR / median")
plt.title("Jet $p_T$ resolution vs epoch")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
plt.plot(epochs, median_val, label="Validation")
plt.axhline(1.0, color="black", ls="--", lw=1)
plt.xlabel("Epoch")
plt.ylabel("Median($p_T^{pred} / p_T^{true}$)")
plt.title("Jet $p_T$ response vs epoch")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```

```python
res_train = np.array(hist["IQR_train"]) / np.array(hist["median_train"])

plt.figure(figsize=(7, 5))
plt.plot(epochs, res_train, label="Train")
plt.plot(epochs, resolution_val, label="Validation")
plt.xlabel("Epoch")
plt.ylabel("IQR / median")
plt.title("Resolution: train vs validation")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```
