```python
# pt regression
jet_p4s = g.reinitialize_p4(zz_data.reco_jet_p4s)
gen_jet_tau_p4s = g.reinitialize_p4(zz_data.gen_jet_tau_p4s)

reco_jet_pt = torch.tensor(ak.to_numpy(jet_p4s.pt), dtype=torch.float32)
gen_tau_pt = torch.tensor(ak.to_numpy(gen_jet_tau_p4s.pt), dtype=torch.float32)

p4_regression_pt = torch.log(gen_tau_pt/reco_jet_pt)
```

```python
# phi
p4_regression_sin_phi = torch.tensor(np.sin(ak.to_numpy(gen_jet_tau_p4s.phi)), dtype=torch.float32)
p4_regression_cos_phi = torch.tensor(np.cos(ak.to_numpy(gen_jet_tau_p4s.phi)), dtype=torch.float32)
```

```python
# mass or energy?
reco_jet_mass = torch.tensor(ak.to_numpy(jet_p4s.mass), dtype=torch.float32)
gen_tau_mass = torch.tensor(ak.to_numpy(gen_jet_tau_p4s.mass), dtype=torch.float32)

p4_regression_mass = torch.log(gen_tau_mass/reco_jet_mass)


reco_jet_energy = torch.tensor(ak.to_numpy(jet_p4s.energy), dtype=torch.float32)
gen_tau_energy = torch.tensor(ak.to_numpy(gen_jet_tau_p4s.energy), dtype=torch.float32)

p4_regression_energy = torch.log(gen_tau_energy/reco_jet_energy)
```
